## Подготовка данных для обучения нейросетей

```python
import torch
from torch.utils.data import Dataset, DataLoader, Subset, random_split
import pandas as pd
```

```python
# Класс Dataset для работы с pandas DataFrame
class TableDataset(Dataset):

    def __init__(self, dataframe):
        data_X = dataframe.drop(columns = 'target')
        data_y = dataframe['target']
        self.X = torch.tensor(data_X.to_numpy(dtype="float32"), dtype=torch.float32)
        self.y = torch.tensor(data_y.to_numpy(), dtype=torch.long)

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```
