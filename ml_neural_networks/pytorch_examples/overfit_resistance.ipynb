# Способы борьбы с переобучением 

## Dropout

**Dropout** -- вспомогательный слой, который с заданной вероятностью отключает случайные нейроны во время обучения.

Идея:
- не позволять модели слишком сильно запоминать обучающие данные

```python
import torch
import torch.nn as nn
```

```python
class SimpleNet(nn.Module):
    def __init__(self):
        super(SimpleNet, self).__init__()
        self.fc1 = nn.Linear(784, 256)
        self.dropout1 = nn.Dropout(p=0.5) # Новый слой
        self.fc2 = nn.Linear(256, 128)
        self.dropout2 = nn.Dropout(p=0.5) # Новый слой
        self.fc3 = nn.Linear(128, 10)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = self.dropout1(x) # Новый слой
        x = torch.relu(self.fc2(x))
        x = self.dropout2(x) # Новый слой
        x = self.fc3(x)
        return x
```

## Регуляризация

Идея:
- штраф за слишком сложную модель (большие веса)

```python
model = SimpleNet()
optimizer = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)
```

## Early stopping

    Ранняя остановка обучения

После каждой эпохи качество модели на валидационной выборке сравнивается с лучшим достигнутым значением. 
Если качество улучшилось, обучение продолжается. Если нет — увеличивается счётчик эпох без улучшения. 
Если этот счётчик превышает заданный порог, обучение прекращается
