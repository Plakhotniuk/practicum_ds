# Основные концепции Airflow

```python
from airflow import DAG
from airflow.operators.empty import EmptyOperator
from airflow.operators.python import PythonOperator
from datetime import datetime
```

```python
def say_hello():
    print('Hello from my function')

dag = DAG('python_dag', 
          description='Python DAG', 
          schedule_interval='*/5 * * * *', 
          start_date=datetime(2022, 1, 1), 
          catchup=False)


dummy_task     = EmptyOperator(task_id='dummy_task', retries=3)
python_task    = PythonOperator(task_id='python_task', python_callable=say_hello)

dummy_task >> python_task 
```

```python
def my_decorator(func):
    def wrapper():
        print("Что-то делаем до вызова функции")
        func()  # Вызов оригинальной функции
        print("Что-то делаем после вызова функции")
    return wrapper

@my_decorator
def say_hello():
    print("Hello, World!")

say_hello()
```

```python
from datetime import datetime
from airflow.decorators import dag, task
from airflow.utils.dates import days_ago

# Определяем DAG с помощью декоратора @dag
@dag(
    dag_id="example_dag_with_decorators",
    schedule_interval="@daily",  # Запускать ежедневно
    start_date=days_ago(1),  # Начать со вчерашнего дня
    catchup=False,  # Не запускать задним числом
)
def example_dag():
    # Определяем задачу с помощью декоратора @task
    @task()
    def extract_data():
        data = {"key": "value"}
        print("Данные извлечены:", data)
        return data

    # Ещё одна задача
    @task()
    def transform_data(data: dict):
        transformed_data = {k.upper(): v.upper() for k, v in data.items()}
        print("Данные преобразованы:", transformed_data)
        return transformed_data

    # Третья задача
    @task()
    def load_data(transformed_data: dict):
        print("Данные загружены:", transformed_data)

    # Определяем зависимости между задачами
    data = extract_data()
    transformed_data = transform_data(data)
    load_data(transformed_data)

# Вызываем функцию DAG, чтобы Airflow её обнаружил
dag = example_dag()
```
